### Harris
#### 基本概念
- 用类似sobel算子算出x和y方向上的梯度变化
- 可以得到一个不规则的椭圆，表示x和y方向上的梯度变化。
- 对这个椭圆进行标准化处理，即得到一个规则椭圆。即得到梯度变化最大方向的变化值(λ1)，和垂直于它的方向的变化值(λ2)
- 当λ1和λ2都很小时，即是一个平坦的面。当λ1和λ2都很大是，即是角
- 可以用λ1和λ2算出一个R值，这个计算式中会用到k的系数
#### cv2.cornerHarris()
- blocksize: 角点检测中指定区域的大小
- ksize: Sobel求导中使用的窗口大小
- k: 计算R的系数
